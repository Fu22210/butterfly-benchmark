```python
import os, getpass, urllib.request, urllib.parse, urllib.error
from collections import Counter
from datetime import datetime, timezone

BBOX_POOL_ID = "6871746"
REVIEW_POOL_ID = "6872092"
TOKEN_FILE = Path("/tmp/crowdsourcing-yandex-tasks.token")

def api_request(method, path, payload=None, params=None):
    token = os.environ.get("YANDEX_TASKS_TOKEN", "").strip()
    if not token and TOKEN_FILE.is_file(): token = TOKEN_FILE.read_text().strip()
    if not token: token = getpass.getpass("OAuth-токен: ").strip()
    if not token: raise ValueError("Токен не задан")
    url = "https://tasks.yandex.ru" + path
    if params: url += "?" + urllib.parse.urlencode(params)
    request = urllib.request.Request(url, method=method,
        data=None if payload is None else json.dumps(payload).encode(),
        headers={"Authorization": "OAuth " + token, "Content-Type": "application/json"})
    try:
        with urllib.request.urlopen(request, timeout=60) as response: return json.load(response)
    except urllib.error.HTTPError as error:
        try:
            details=json.loads(error.read().decode('utf-8'))
        except (ValueError,UnicodeDecodeError):
            details={}
        fields=details.get('payload',{})
        validation={str(k):str(v.get('code','invalid')) for k,v in fields.items() if isinstance(v,dict)} if isinstance(fields,dict) else {}
        message='; '.join(k+': '+v for k,v in validation.items())[:1500]
        failure=RuntimeError(f"API HTTP {error.code}" + (" — "+message if message else "; проверьте настройки и доступ"))
        failure.status_code=error.code
        failure.validation_errors=validation
        raise failure from None
    except urllib.error.URLError:
        raise RuntimeError("Сбой соединения с API; запросы записи автоматически не повторяются") from None

def api_list(resource, **params):
    params = dict(params, sort="id", limit=100)
    items, seen = [], set()
    while True:
        page = api_request("GET", "/api/v1/" + resource, params=params)
        batch = page["items"]
        for item in batch:
            if item["id"] in seen: raise ValueError("Повтор страницы API")
            seen.add(item["id"]); items.append(item)
        if not page.get("has_more"): return items
        if not batch: raise ValueError("Пустая страница API с has_more")
        params["id_gt"] = batch[-1]["id"]

bbox_pool = api_request("GET", "/api/v1/pools/" + BBOX_POOL_ID)
if str(bbox_pool["project_id"]) != "11941": raise ValueError("Неверный bbox-проект")
assignments = api_list("assignments", pool_id=BBOX_POOL_ID)
fields = ["ASSIGNMENT:worker_id", "ASSIGNMENT:assignment_id", "ASSIGNMENT:status",
          "ASSIGNMENT:submitted", "INPUT:photo_id", "INPUT:image", "OUTPUT:status", "OUTPUT:result"]
rows = []
for a in assignments:
    if str(a["pool_id"]) != BBOX_POOL_ID: raise ValueError("Ответ из другого пула")
    if a["status"] not in ("ACCEPTED", "SUBMITTED"): continue
    tasks, solutions = a["tasks"], a["solutions"]
    if len(tasks) != len(solutions): raise ValueError("Не совпадает число задач и ответов")
    if any(s.get("task_id") for s in solutions):
        by_id = {s["task_id"]: s for s in solutions}
        if len(by_id) != len(solutions): raise ValueError("Повтор task_id")
        solutions = [by_id[t["id"]] for t in tasks]
    for task, solution in zip(tasks, solutions):
        inp, out = task["input_values"], solution["output_values"]
        rows.append({"ASSIGNMENT:worker_id": a["user_id"], "ASSIGNMENT:assignment_id": a["id"],
                     "ASSIGNMENT:status": a["status"], "ASSIGNMENT:submitted": a.get("submitted", ""),
                     "INPUT:photo_id": inp["photo_id"], "INPUT:image": inp["image"],
                     "OUTPUT:status": out.get("status", ""), "OUTPUT:result": json.dumps(out.get("result"))})
# Снимки сохраняются отдельно; рабочая выгрузка заменяется только после полной загрузки.
stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
archive = BASE / "snapshots"; archive.mkdir(exist_ok=True)
with (archive / ("bbox_" + stamp + ".tsv")).open("w", encoding="utf-8-sig", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fields, delimiter="\t"); writer.writeheader(); writer.writerows(rows)
BBOX_EXPORT.write_bytes((archive / ("bbox_" + stamp + ".tsv")).read_bytes())
print("Назначения:", dict(Counter(a["status"] for a in assignments)))
print("Скачано завершённых ответов:", len(rows))
```

```python
def boxes_ltwh(raw):
    value = json.loads(raw) if isinstance(raw, str) else raw
    if not isinstance(value, list):
        raise ValueError('Рамки должны быть массивом')
    result = []
    for b in value:
        if not isinstance(b, dict):
            raise ValueError('Рамка должна быть объектом')
        if b.get('type') == 'rectangle':
            a, z = b['data']['p1'], b['data']['p2']
            l, t = min(a['x'], z['x']), min(a['y'], z['y'])
            w, h = abs(a['x']-z['x']), abs(a['y']-z['y'])
        else:
            l, t, w, h = (b[k] for k in ('left','top','width','height'))
        import math
        if any(isinstance(x, bool) or not isinstance(x, (int,float)) or not math.isfinite(x) for x in (l,t,w,h)) or min(l,t)<0 or min(w,h)<=0 or l+w>1.000001 or t+h>1.000001:
            raise ValueError('Некорректная геометрия')
        result.append(dict(left=l, top=t, width=w, height=h))
    return result

previous_map_path = OUT / 'private_source_map.json'
if previous_map_path.exists():
    import uuid
    map_archive = BASE / 'snapshots'; map_archive.mkdir(exist_ok=True)
    (map_archive / ('source_map_' + uuid.uuid4().hex + '.json')).write_bytes(previous_map_path.read_bytes())
mapping = []
authors = set()
if BBOX_EXPORT.exists():
    grouped = defaultdict(list)
    for r in read_tsv(BBOX_EXPORT):
        author = pick(r, 'ASSIGNMENT:worker_id', 'worker_id')
        if author: authors.add(author)
        state = pick(r, 'ASSIGNMENT:status', 'assignment_status').upper()
        if state not in ('ACCEPTED', 'SUBMITTED'): continue
        grouped[pick(r, 'INPUT:photo_id', 'photo_id')].append(r)
    tasks, issues = [], []
    for photo, rows in grouped.items():
        try:
            if len(rows) != 1: raise ValueError('Несколько ответов: выберите окончательный явно')
            r = rows[0]
            author = pick(r, 'ASSIGNMENT:worker_id', 'worker_id')
            assignment = pick(r, 'ASSIGNMENT:assignment_id', 'assignment_id')
            status = pick(r, 'OUTPUT:status', 'status')
            if not author or not assignment: raise ValueError('Нет автора или ID ответа')
            boxes = boxes_ltwh(pick(r, 'OUTPUT:result', 'result'))
            if status not in ('boxes','none') or (status == 'none') != (len(boxes)==0): raise ValueError('Ошибка загрузки или противоречивый ответ')
            image = pick(r, 'INPUT:image', 'image')
            if not image.startswith(('https://','http://')): raise ValueError('Нет URL изображения')
            rid = 'review-' + str(photo)
            tasks.append({'INPUT:review_id':rid, 'INPUT:photo_id':int(photo), 'INPUT:image':image, 'INPUT:boxes':boxes})
            mapping.append(dict(review_id=rid, photo_id=int(photo), author_id=author, assignment_id=assignment, image=image, boxes=boxes))
        except (ValueError, KeyError, TypeError) as e:
            issues.append(dict(photo_id=photo, issue=str(e)))
    write_tsv('production_tasks.tsv', tasks, input_fields)
    write_tsv('source_issues.tsv', issues, ['photo_id','issue'])
    write_tsv('authors_to_exclude.tsv', [{'worker_id':x,'value':1} for x in sorted(authors)], ['worker_id','value'])
    (OUT/'private_source_map.json').write_text(json.dumps(mapping, ensure_ascii=False, indent=2))
    # API-level exclusion is not an input label: the platform enforces unavailable_for.
    state_path=BASE/'api_state.json'
    pool_id=json.loads(state_path.read_text())['main_pool_id'] if state_path.exists() else None
    api_tasks=[dict(pool_id=pool_id,input_values={k.removeprefix('INPUT:'):v for k,v in t.items()},overlap=1,infinite_overlap=False,unavailable_for=sorted(authors)) for t in tasks]
    (OUT/'production.tasks.json').write_text(json.dumps(api_tasks,ensure_ascii=False,indent=2))
    print('Готово:',len(tasks),'фото; требуют решения:',len(issues),'авторов исключить:',len(authors))
    print('Перед запуском проверьте покрытие всех 400 исходных photo_id, включая отсутствующие в выгрузке.')
else:
    print('Выгрузки рамок ещё нет — реальные задания ревью пока не сформированы.')
```

## Загрузка в основной review-пул
Проверяем существующие review_id, пропускаем уже загруженные работы и останавливаемся при изменении исходного ответа. Контроли не меняются. Пул не открывается. Запускайте загрузку только в одном экземпляре ноутбука.

При сетевом сбое запись сохраняется как sending: повторный запуск сначала ищет задание на сервере; если его нет, останавливается для проверки, а не создаёт возможный дубль.


```python
def plan_review_upload(local_tasks, remote_tasks):
    remote = {}
    for task in remote_tasks:
        rid = task["input_values"].get("review_id")
        if rid in remote: raise ValueError("Дубли review_id на платформе")
        remote[rid] = task
    pending, existing, seen = [], [], set()
    for task in local_tasks:
        rid = task["input_values"]["review_id"]
        if rid in seen: raise ValueError("Повтор review_id в локальном плане")
        seen.add(rid)
        if str(task["pool_id"]) != REVIEW_POOL_ID or not task.get("unavailable_for"):
            raise ValueError("Неверный пул или отсутствует запрет самопроверки")
        if rid not in remote:
            pending.append(task); continue
        saved = remote[rid]
        if saved["input_values"] != task["input_values"] or saved.get("known_solutions"):
            raise ValueError("Ранее загруженный ответ изменился; нужен отдельный цикл исправления")
        # Автор конкретного ответа обязательно должен оставаться исключённым.
        author = next(r["author_id"] for r in mapping if r["review_id"] == rid)
        if author not in saved.get("unavailable_for", []):
            raise ValueError("В существующем задании отсутствует запрет самопроверки")
        existing.append(saved)
    return pending, existing

review_pool = api_request("GET", "/api/v1/pools/" + REVIEW_POOL_ID)
if str(review_pool["project_id"]) != "11946": raise ValueError("Неверный review-проект")
local_tasks = json.loads((OUT / "production.tasks.json").read_text())
remote_tasks = api_list("tasks", pool_id=REVIEW_POOL_ID)
pending_tasks, existing_tasks = plan_review_upload(local_tasks, remote_tasks)
print("Подготовлено:", len(local_tasks), "Уже загружено:", len(existing_tasks), "Новых:", len(pending_tasks))
print("Статус review:", review_pool["status"], "Цена страницы:", review_pool["reward_per_assignment"])
```

```python
UPLOAD_REVIEW = False  # True — загрузить новые задания; пул останется закрытым

def upload_new_reviews(local_tasks):
    pool = api_request("GET", "/api/v1/pools/" + REVIEW_POOL_ID)
    if pool["status"] != "CLOSED": raise RuntimeError("Перед загрузкой закройте review-пул")
    pending, existing = plan_review_upload(local_tasks, api_list("tasks", pool_id=REVIEW_POOL_ID))
    journal = OUT / "review_upload_journal.json"
    history = json.loads(journal.read_text()) if journal.exists() else {}
    for task in pending:
        rid = task["input_values"]["review_id"]
        if rid in history:
            raise RuntimeError("Ранее отправленное задание не найдено на сервере; проверьте журнал перед повторной загрузкой")
    def checkpoint():
        temp = journal.with_suffix(".tmp")
        temp.write_text(json.dumps(history, ensure_ascii=False, indent=2)); temp.replace(journal)
    for task in pending:
        rid = task["input_values"]["review_id"]
        history[rid] = {"state": "sending", "time": datetime.now(timezone.utc).isoformat()}
        checkpoint()
        saved = api_request("POST", "/api/v1/tasks", task,
                            params={"async_mode": "false", "open_pool": "false"})
        checked = api_request("GET", "/api/v1/tasks/" + saved["id"])
        if checked["input_values"] != task["input_values"] or not set(task["unavailable_for"]).issubset(checked.get("unavailable_for", [])):
            raise RuntimeError("Загруженное задание не прошло проверку")
        history[rid] = {"state": "verified", "task_id": checked["id"],
                        "time": datetime.now(timezone.utc).isoformat()}
        checkpoint()
    remaining, existing = plan_review_upload(local_tasks, api_list("tasks", pool_id=REVIEW_POOL_ID))
    if remaining: raise RuntimeError("Не все задания появились на платформе")
    print("Подтверждено на платформе:", len(existing), "Новых загружено:", len(pending))

if UPLOAD_REVIEW:
    upload_new_reviews(local_tasks)
else:
    print("Предпросмотр. Для загрузки установите UPLOAD_REVIEW = True и повторите эту ячейку.")
validation = api_request("GET", "/api/new/requester/pools/" + REVIEW_POOL_ID + "/validate")
(OUT / "review_launch_validation.json").write_text(json.dumps(validation, ensure_ascii=False, indent=2))
print("Активные предупреждения:", [(v["type"], v["blocker"]) for v in validation if v["active"]])
```

## Оценка экзамена проверяющих
Бесплатный экзамен 6872920: только 9/10. Платформа автоматически записывает фактический процент в навык 21849 и допуск 21819: 100 за 9/10, иначе 0 после всех 10 ответов. Ноутбук проверяет то же правило по ID пула. Старый экзамен 6872091 сохраняет 8/10, минимум 4/6 reject и 3/4 accept; ранее выданные допуски не меняются.


```python
REVIEW_EXAM_POOL_ID = "6872920"
LEGACY_REVIEW_EXAM_POOL_ID = "6872091"
REVIEW_SKILL_ID = "21819"

def fetch_review_rows(pool_id):
    pool = api_request("GET", "/api/v1/pools/" + pool_id)
    if str(pool["project_id"]) != "11946": raise ValueError("Неверный review-проект")
    result = []
    assignments = api_list("assignments", pool_id=pool_id)
    for a in assignments:
        if str(a["pool_id"]) != pool_id: raise ValueError("Ответ из другого пула")
        if a["status"] not in ("ACCEPTED", "SUBMITTED"): continue
        tasks, solutions = a["tasks"], a["solutions"]
        if len(tasks) != len(solutions): raise ValueError("Количество ответов не совпадает с заданиями")
        if any(s.get("task_id") for s in solutions):
            by_id = {s["task_id"]:s for s in solutions}
            if len(by_id) != len(solutions): raise ValueError("Повтор task_id")
            solutions = [by_id[t["id"]] for t in tasks]
        for task, solution in zip(tasks, solutions):
            inp, out = task["input_values"], solution["output_values"]
            result.append({"ASSIGNMENT:pool_id":pool_id, "ASSIGNMENT:worker_id":a["user_id"], "ASSIGNMENT:assignment_id":a["id"],
                "ASSIGNMENT:status":a["status"], "ASSIGNMENT:submitted":a.get("submitted", ""),
                "INPUT:review_id":inp["review_id"], "INPUT:photo_id":inp["photo_id"],
                "OUTPUT:verdict":out.get("verdict", ""), "OUTPUT:reasons":out.get("reasons", []),
                "OUTPUT:comment":out.get("comment", "")})
    print("Пул", pool_id, "Назначения:", dict(Counter(a["status"] for a in assignments)))
    return result

def save_review_export(path, rows):
    fields = ["ASSIGNMENT:pool_id", "ASSIGNMENT:worker_id", "ASSIGNMENT:assignment_id", "ASSIGNMENT:status",
        "ASSIGNMENT:submitted", "INPUT:review_id", "INPUT:photo_id", "OUTPUT:verdict", "OUTPUT:reasons", "OUTPUT:comment"]
    snapshots = BASE / "snapshots"; snapshots.mkdir(exist_ok=True)
    name = path.stem + "_" + datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S%fZ") + ".tsv"
    snapshot_path = snapshots / name
    with snapshot_path.open("w", encoding="utf-8-sig", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fields, delimiter="\t"); writer.writeheader()
        writer.writerows({k:json.dumps(v, ensure_ascii=False) if isinstance(v,(list,dict)) else v for k,v in r.items()} for r in rows)
    path.write_bytes(snapshot_path.read_bytes())

review_exam_rows = fetch_review_rows(LEGACY_REVIEW_EXAM_POOL_ID) + fetch_review_rows(REVIEW_EXAM_POOL_ID)
save_review_export(EXAM_EXPORT, review_exam_rows)
# Исключить всех участников bbox-экзамена с назначенным навыком, включая не прошедших.
bbox_skill_holders = {r["user_id"] for r in api_list("user-skills", skill_id="21809")}
excluded_review_workers = set(authors) | bbox_skill_holders
print("Строк экзамена review:", len(review_exam_rows))
```

```python
def score_exam(rows, key, excluded=()):
    grouped = defaultdict(list)
    for r in rows:
        if pick(r,'ASSIGNMENT:status','assignment_status').upper() not in ('SUBMITTED','ACCEPTED'): continue
        worker = pick(r,'ASSIGNMENT:worker_id','worker_id')
        rid = pick(r,'INPUT:review_id','review_id')
        if not worker or rid not in key: raise ValueError('Неизвестное задание или пустой worker_id')
        grouped[worker].append((rid,pick(r,'OUTPUT:verdict','verdict')))
    perfect_workers = {pick(r,'ASSIGNMENT:worker_id','worker_id') for r in rows
                       if r.get('ASSIGNMENT:pool_id') == '6872920'}
    result=[]
    for worker, answers in grouped.items():
        ids=[x[0] for x in answers]
        if worker in excluded: decision='author_excluded'
        elif len(ids)!=len(set(ids)): decision='duplicate_answers'
        elif set(ids)!=set(key): decision='incomplete'
        elif any(v=='image_error' for _,v in answers): decision='technical_issue'
        elif any(v not in ('accept','reject','uncertain') for _,v in answers): decision='invalid_verdict'
        else: decision='graded'
        good=sum(v==key[rid] for rid,v in answers)
        rejected=sum(v==key[rid]=='reject' for rid,v in answers)
        accepted=sum(v==key[rid]=='accept' for rid,v in answers)
        passed=decision=='graded' and (good==10 if worker in perfect_workers else good>=8 and rejected>=4 and accepted>=3)
        result.append(dict(worker_id=worker, correct=good, correct_reject=rejected, correct_accept=accepted, decision=('passed' if passed else 'failed') if decision=='graded' else decision, skill_value=good*10 if passed else (0 if decision=='graded' else '')))
    return result

if samples is not None and EXAM_EXPORT.exists():
    key={s['review_id']:s['verdict'] for s in samples if s['split']=='exam'}
    if list(key.values()).count('accept')!=4 or list(key.values()).count('reject')!=6: raise ValueError('Изменился баланс: пересмотрите пороги')
    grades=score_exam(read_tsv(EXAM_EXPORT),key,globals().get('excluded_review_workers', authors))
    write_tsv('review_exam_scores.tsv',grades,['worker_id','correct','correct_reject','correct_accept','decision','skill_value'])
    print(grades)
```

## Назначение экзаменационного навыка ревьюерам
Прошедшим — 80–100, завершившим и не прошедшим — 0. Неполные, повторные и технически проблемные ответы, а также авторы bbox исключаются. Сначала проверьте таблицу оценок. Затем включите APPLY_REVIEW_SKILLS в следующей ячейке. Повторный запуск не перезаписывает совпадающие значения.


```python
def review_skill_plan(exam_rows, excluded):
    if samples is None: raise ValueError("Нет проверенных эталонов")
    key = {s["review_id"]:s["verdict"] for s in samples if s["split"] == "exam"}
    if len(key) != 10 or list(key.values()).count("reject") != 6:
        raise ValueError("Изменился состав экзамена")
    grades = score_exam(exam_rows, key, excluded)
    plan = []
    for r in grades:
        if r["decision"] not in ("passed", "failed"): continue
        value = r["skill_value"]
        if value not in (0,80,90,100): raise ValueError("Некорректная оценка")
        plan.append({"skill_id": REVIEW_SKILL_ID, "user_id": r["worker_id"], "value": value})
    return sorted(plan, key=lambda r:r["user_id"])

review_skill_assignments = review_skill_plan(read_tsv(EXAM_EXPORT), excluded_review_workers)
print("К назначению навыка:", len(review_skill_assignments))
print("Распределение:", dict(Counter(r["value"] for r in review_skill_assignments)))
for row in review_skill_assignments: print(row["user_id"], "→", row["value"])
write_tsv("review_skill_assignments.tsv", review_skill_assignments, ["skill_id", "user_id", "value"])
```

```python
APPLY_REVIEW_SKILLS = False  # True — назначить навык 21819 и проверить сохранение

def get_review_skill(user_id):
    rows = api_list("user-skills", skill_id=REVIEW_SKILL_ID, user_id=user_id)
    if len(rows) > 1: raise ValueError("Несколько значений одного навыка")
    if not rows: return None
    r = rows[0]
    if str(r["skill_id"]) != REVIEW_SKILL_ID or r["user_id"] != user_id:
        raise ValueError("Ответ API не соответствует запросу")
    return r

def skill_matches(record, value):
    return record is not None and float(record.get("exact_value", record.get("value"))) == value

def assign_review_skills(plan):
    counts = Counter()
    for payload in plan:
        before = get_review_skill(payload["user_id"])
        action = "already_set" if skill_matches(before, payload["value"]) else "updated"
        if action == "updated": api_request("PUT", "/api/v1/user-skills", payload)
        after = get_review_skill(payload["user_id"])
        if not skill_matches(after, payload["value"]): raise RuntimeError("Сохранение навыка не подтверждено")
        with (OUT / "review_skill_log.jsonl").open("a", encoding="utf-8") as f:
            f.write(json.dumps({**payload, "action":action, "verified":True,
                "time":datetime.now(timezone.utc).isoformat()}, ensure_ascii=False) + "\n")
        counts[action] += 1
    return dict(counts)

if APPLY_REVIEW_SKILLS:
    fresh_rows = fetch_review_rows(LEGACY_REVIEW_EXAM_POOL_ID) + fetch_review_rows(REVIEW_EXAM_POOL_ID)
    fresh_excluded = set(authors) | {r["user_id"] for r in api_list("user-skills", skill_id="21809")}
    fresh_plan = review_skill_plan(fresh_rows, fresh_excluded)
    if fresh_plan != review_skill_assignments:
        raise RuntimeError("Экзамен или исключения изменились. Повторите загрузку и оценивание.")
    print("Назначение:", assign_review_skills(fresh_plan))
else:
    print("Навыки не изменены. Для записи установите APPLY_REVIEW_SKILLS = True.")
```

## Контроли, квалификация проверяющих и golden
Контрольные строки отделяются от основных. Нужны подтверждённые controls.verified.json, результат полного экзамена и минимум 5 контролей с точностью ≥80%. Если проверяющий когда-либо опускался ниже порога после 5 ответов в текущей выгрузке, его история уходит на пересмотр. Выгрузка должна охватывать весь пул и содержать ASSIGNMENT:submitted. На коротком пилоте недостаток контролей означает ручной аудит команды, а не автоматическое принятие.


```python
# Полная история: и реальные работы, и скрытые контроли, без обрезки по дате.
review_rows = fetch_review_rows(REVIEW_POOL_ID)
save_review_export(REVIEW_EXPORT, review_rows)
print("Скачано строк основного review:", len(review_rows))
```

```python
def finalize_reviews(rows, source, control_key, passed_workers, excluded=(), controls_confirmed=False):
    """Separate controls from production; never silently accept unqualified reviewers."""
    active = [r for r in rows if pick(r,'ASSIGNMENT:status','assignment_status').upper() in ('SUBMITTED','ACCEPTED')]
    unknown = {pick(r,'INPUT:review_id','review_id') for r in active} - set(source) - set(control_key)
    if unknown: raise ValueError('В выгрузке неизвестные задания')
    control_rows, decisions = defaultdict(list), defaultdict(list)
    for r in active:
        rid=pick(r,'INPUT:review_id','review_id')
        if rid in control_key: control_rows[pick(r,'ASSIGNMENT:worker_id','worker_id')].append(r)
        else: decisions[rid].append(r)
    stats={}
    for worker, answers in control_rows.items():
        ids=[pick(r,'INPUT:review_id','review_id') for r in answers]
        missing_time=any(not pick(r,'ASSIGNMENT:submitted','submitted') for r in answers)
        answers=sorted(answers,key=lambda r:pick(r,'ASSIGNMENT:submitted','submitted'))
        good=0; failed_earlier=False
        for i,r in enumerate(answers,1):
            good += pick(r,'OUTPUT:verdict','verdict') == control_key[pick(r,'INPUT:review_id','review_id')]
            if i>=5 and good/i<.8: failed_earlier=True
        decision=('unverified_controls' if not controls_confirmed else
                  'missing_worker' if not worker else
                  'duplicate_control_answers' if len(ids)!=len(set(ids)) else
                  'missing_control_time' if missing_time else
                  'insufficient_controls' if len(answers)<5 else
                  'failed_controls' if failed_earlier else 'passed')
        stats[worker]=dict(worker_id=worker,control_count=len(answers),correct=good,accuracy=good/len(answers),decision=decision)
    excluded=set(excluded)|{s['author_id'] for s in source.values()}
    golden,queue=[],[]
    for rid,s in source.items():
        rr=decisions.get(rid,[])
        reason='missing_review' if not rr else 'multiple_reviews' if len(rr)!=1 else ''
        if not reason:
            r=rr[0];worker=pick(r,'ASSIGNMENT:worker_id','worker_id');verdict=pick(r,'OUTPUT:verdict','verdict')
            if not worker or worker in excluded:reason='author_excluded_or_missing_worker'
            elif worker in s.get('previous_reviewers', []):reason='previous_reviewer_excluded'
            elif worker not in passed_workers:reason='exam_not_passed_or_unknown'
            elif stats.get(worker,{}).get('decision')!='passed':reason=stats.get(worker,{}).get('decision','no_control_answers')
            elif not pick(r,'ASSIGNMENT:assignment_id','assignment_id'):reason='missing_review_assignment'
            elif verdict!='accept':reason=verdict or 'missing_verdict'
            else:golden.append(dict(s,reviewer_id=worker,review_assignment_id=pick(r,'ASSIGNMENT:assignment_id','assignment_id'),verdict=verdict))
        if reason:queue.append(dict(review_id=rid,photo_id=s['photo_id'],issue=reason,raw_reviews=rr))
    return golden,queue,list(stats.values())

if REVIEW_EXPORT.exists() and (OUT/'private_source_map.json').exists():
    source={s['review_id']:s for s in json.loads((OUT/'private_source_map.json').read_text())}
    correction_map = BASE/'corrections'/'review_source_map.json'
    if correction_map.exists():
        for item in json.loads(correction_map.read_text()):
            source[item['review_id']] = item
    controls_file=BASE/'controls.verified.json'
    controls_confirmed=controls_file.exists()
    controls=json.loads((controls_file if controls_confirmed else BASE/'controls.to_verify.json').read_text())
    controls_confirmed=controls_confirmed and len(controls)==14 and all(s.get('human_verified') is True for s in controls)
    control_key={s['review_id']:s['verdict'] for s in controls}
    if len(control_key)!=len(controls):raise ValueError('Повторяющиеся контрольные задания')
    grades_file=OUT/'review_exam_scores.tsv'
    passed_workers={r['worker_id'] for r in read_tsv(grades_file) if r['decision']=='passed'} if grades_file.exists() else set()
    golden,queue,stats=finalize_reviews(read_tsv(REVIEW_EXPORT),source,control_key,passed_workers,authors,controls_confirmed)
    write_tsv('control_scores.tsv',stats,['worker_id','control_count','correct','accuracy','decision'])
    (OUT/'accepted_originals.json').write_text(json.dumps(golden,ensure_ascii=False,indent=2))
    (OUT/'correction_queue.json').write_text(json.dumps(queue,ensure_ascii=False,indent=2))
    print('Приняты:',len(golden),'требуют работы:',len(queue))
    print('Если контролей пока меньше 5, работы остаются в очереди до ручного аудита команды. Это ожидаемо на коротком пилоте.')
```

По решению заказчика переразмечаем 61 отклонённое фото, 21 расхождение, 11 случаев недостаточного контроля и 2 ошибки загрузки. Последние две ссылки проверены: HTTP 200, изображения читаются. Причина сохраняется для аудита и не показывается исполнителю. Допуск — новый бесплатный экзамен 9/10; reserved_for не используется. После новой разметки требуется независимый review.


```python
# Цикл исправлений v2. На платформу передаются только фото и новые рамки, без причин.
import copy
CORRECTIONS = BASE / 'corrections'
CORRECTIONS.mkdir(exist_ok=True)
CORRECTION_STATE = CORRECTIONS / 'state.json'

def save_correction_state(state):
    tmp = CORRECTION_STATE.with_suffix('.tmp')
    tmp.write_text(json.dumps(state, ensure_ascii=False, indent=2)); tmp.replace(CORRECTION_STATE)

def load_correction_state():
    return json.loads(CORRECTION_STATE.read_text()) if CORRECTION_STATE.exists() else {}

def correction_candidates(queue, original_source):
    """User decision: relabel rejects, disagreements, insufficient controls and verified image errors."""
    result = []
    for item in queue:
        if item.get('issue') not in ('reject','disagreement_or_uncertain','insufficient_controls','image_error'): continue
        rid = item['review_id']
        if rid.endswith('-v2') and rid[:-3] in original_source: continue
        if rid not in original_source: raise ValueError('Неизвестный исходный ответ')
        source = original_source[rid]
        if source.get('attempt', 1) != 1: continue  # повторный reject -> ручной разбор
        reviewers = sorted({pick(r, 'ASSIGNMENT:worker_id', 'worker_id') for r in item.get('raw_reviews', [])}
                           - {''})
        if not reviewers: raise ValueError('Неизвестен проверяющий отклонённого ответа')
        result.append(dict(photo_id=source['photo_id'], image=source['image'],
            previous_review_id=rid, original_author_id=source['author_id'], previous_reviewers=reviewers, reason=item['issue']))
    if len({r['photo_id'] for r in result}) != len(result):
        raise ValueError('Несколько отклонений одного фото: нужен разбор')
    return sorted(result, key=lambda r:r['photo_id'])

def task_key(task):
    inp = task['input_values']
    return str(inp.get('review_id', inp.get('photo_id')))

def upload_correction_tasks(pool_id, tasks, journal_name):
    pool = api_request('GET', '/api/v1/pools/' + pool_id)
    if pool['status'] != 'CLOSED': raise RuntimeError('Загрузка разрешена только в закрытый пул')
    journal = CORRECTIONS / journal_name
    history = json.loads(journal.read_text()) if journal.exists() else {}
    remote = api_list('tasks', pool_id=pool_id)
    existing = {}
    for t in remote:
        key = task_key(t)
        if key in existing: raise ValueError('Дубли заданий на платформе')
        existing[key] = t
    if len({task_key(t) for t in tasks}) != len(tasks): raise ValueError('Дубли в плане')
    def check(expected, saved):
        if saved['input_values'] != expected['input_values'] or str(saved['pool_id']) != pool_id:
            raise ValueError('Содержимое существующего задания отличается')
        if set(expected.get('reserved_for', [])) != set(saved.get('reserved_for', [])):
            raise ValueError('Не сохранён список допущенных исполнителей')
        if saved.get('overlap') != expected.get('overlap') or saved.get('infinite_overlap') != expected.get('infinite_overlap'):
            raise ValueError('Не сохранено перекрытие')
        if saved.get('known_solutions',[]) != expected.get('known_solutions',[]):
            raise ValueError('Не сохранены эталонные ответы')
        if not set(expected.get('unavailable_for', [])).issubset(saved.get('unavailable_for', [])):
            raise ValueError('Не сохранены исключения исполнителей')
    # Validate the whole plan before the first write.
    for t in tasks:
        if str(t['pool_id']) != pool_id: raise ValueError('Неверный пул')
        key = task_key(t)
        if key in existing: check(t, existing[key])
        elif key in history: raise RuntimeError('Неясный результат прошлой записи; проверьте журнал, не повторяйте POST')
    new_count = 0
    for t in tasks:
        key = task_key(t)
        if key in existing: continue
        history[key] = {'state':'sending'}
        journal.write_text(json.dumps(history, ensure_ascii=False, indent=2))
        saved = api_request('POST', '/api/v1/tasks', t, params={'async_mode':'false', 'open_pool':'false'})
        checked = api_request('GET', '/api/v1/tasks/' + saved['id']); check(t, checked)
        history[key] = {'state':'verified', 'task_id':checked['id']}
        journal.write_text(json.dumps(history, ensure_ascii=False, indent=2)); new_count += 1
    return new_count

def create_correction_pool(batch):
    state = load_correction_state()
    if state.get('pool_id'): return state
    if state.get('creation_pending'):
        raise RuntimeError('Предыдущий POST пула мог выполниться. Проверьте аккаунт и state.json, не создавайте дубль')
    if not batch: return state
    template = api_request('GET', '/api/v1/pools/' + BBOX_POOL_ID)
    if str(template['project_id']) != '11941': raise ValueError('Неверный bbox-проект')
    # Сохраняем текущую цену и допуск bbox. Одна фотография, обычная разметка с нуля.
    fields = ['project_id','will_expire','reward_per_assignment','assignment_max_duration_seconds',
              'auto_accept_solutions','auto_accept_period_day','filter','quality_control',
              'may_contain_adult_content']
    payload = {k:copy.deepcopy(template[k]) for k in fields if k in template}
    exam_state_path = CORRECTIONS/'free_exam_state.json'
    if not exam_state_path.exists(): raise ValueError('Сначала создайте бесплатный bbox-экзамен')
    exam_state = json.loads(exam_state_path.read_text())
    payload['filter']={'and':[{'category':'profile','key':'languages','operator':'IN','value':'RU'},
                             {'category':'skill','key':exam_state['score_skill_id'],'operator':'GTE','value':90},
                             {'category':'skill','key':'21819','operator':'EQ'}]}
    for cfg in payload.get('quality_control',{}).get('configs',[]): cfg['collector_config'].pop('uuid',None)
    payload.update(private_name='Переразметка 95 фото — попытка 2',
        defaults={'default_overlap_for_new_tasks':1,'default_overlap_for_new_task_suites':1},
        mixer_config={'real_tasks_count':1,'golden_tasks_count':0,'training_tasks_count':0,
                      'min_real_tasks_count':1,'force_last_assignment':True})
    state = {'creation_pending':True, 'batch':batch, 'attempt':2, 'created_at':datetime.now(timezone.utc).isoformat()}
    save_correction_state(state)
    saved = api_request('POST', '/api/v1/pools', payload)
    state.update(pool_id=str(saved['id']), creation_pending=False); save_correction_state(state)
    checked = api_request('GET', '/api/v1/pools/' + state['pool_id'])
    if checked['status'] != 'CLOSED': raise RuntimeError('Новый пул неожиданно открыт')
    return state

def corrected_answers(assignments, state):
    wanted = {r['photo_id']:r for r in state.get('batch', [])}
    grouped = defaultdict(list)
    for a in assignments:
        if str(a['pool_id']) != state['pool_id']: raise ValueError('Ответ из другого пула исправлений')
        if a['status'] not in ('SUBMITTED','ACCEPTED'): continue
        tasks, solutions = a['tasks'], a['solutions']
        if len(tasks) != len(solutions): raise ValueError('Число задач и ответов не совпадает')
        if any(s.get('task_id') for s in solutions):
            by_id = {s['task_id']:s for s in solutions}
            if len(by_id) != len(solutions): raise ValueError('Повтор task_id')
            solutions = [by_id[t['id']] for t in tasks]
        for task, sol in zip(tasks, solutions):
            photo = int(task['input_values']['photo_id'])
            if photo not in wanted: raise ValueError('Неизвестное фото исправлений')
            grouped[photo].append((a, task, sol))
    records, issues = [], []
    for photo, old in wanted.items():
        answers = grouped.get(photo, [])
        if len(answers) != 1:
            issues.append({'photo_id':photo,'issue':'missing_correction' if not answers else 'multiple_corrections'}); continue
        a, task, sol = answers[0]
        try:
            if task['input_values']['image'] != old['image']: raise ValueError('Изменился URL')
            output = sol['output_values']; boxes = boxes_ltwh(output.get('result'))
            if output.get('status') not in ('none','boxes') or (output['status']=='none') != (len(boxes)==0):
                raise ValueError('Ошибка загрузки или противоречивый ответ')
            if not a.get('user_id') or not a.get('id'): raise ValueError('Нет автора или назначения')
            records.append(dict(review_id='review-'+str(photo)+'-v2', photo_id=photo, image=old['image'], boxes=boxes,
                author_id=a['user_id'], assignment_id=a['id'], attempt=2,
                previous_review_id=old['previous_review_id'], original_author_id=old['original_author_id'],
                previous_reviewers=old['previous_reviewers']))
        except (ValueError, TypeError, KeyError) as error:
            issues.append({'photo_id':photo,'issue':str(error)})
    return records, issues

def correction_review_payloads(records, excluded):
    return [dict(pool_id=REVIEW_POOL_ID,
        input_values={k:r[k] for k in ('review_id','photo_id','image','boxes')},
        overlap=1, infinite_overlap=False,
        unavailable_for=sorted(set(excluded)|{r['author_id'],r['original_author_id']}|set(r['previous_reviewers'])))
        for r in records]

def select_latest_golden(accepted, sources):
    latest = {}
    for r in sources.values():
        old = latest.get(r['photo_id'])
        if old is None or r.get('attempt',1) > old.get('attempt',1): latest[r['photo_id']] = r
    return [r for r in accepted if latest[r['photo_id']]['review_id'] == r['review_id']]
```

## Повторный review ненадёжных проверок — перекрытие 2
Берём только failed_controls, а не все непринятые работы. Исходные рамки сохраняются. Два разных надёжных голоса: accept+accept → принять, reject+reject → переразметка, остальные случаи → ручной разбор.

Контроли используются только для тех кандидатов, которые их ещё не видели. Кандидаты, видевшие почти весь банк, не включаются: нужно минимум 5 незнакомых эталонов. В новом пуле первые 5 заданий контрольные, далее каждое десятое. Оценка учитывает прошлую историю и минимум 5 новых контролей. При нехватке кандидатов нужно набрать новых через бесплатный экзамен либо подготовить новый подтверждённый банк контролей.


```python
# Повторная проверка исходных рамок; это не переразметка.
REREVIEW = BASE / 'rereview'; REREVIEW.mkdir(exist_ok=True)
REREVIEW_STATE = REREVIEW / 'state.json'

def read_rereview_state():
    return json.loads(REREVIEW_STATE.read_text()) if REREVIEW_STATE.exists() else {}

def write_rereview_state(state):
    temp = REREVIEW_STATE.with_suffix('.tmp')
    temp.write_text(json.dumps(state,ensure_ascii=False,indent=2)); temp.replace(REREVIEW_STATE)

def rereview_plan(source, queue, first_rows, controls, exam_passed, excluded, control_stats):
    bad = {s['worker_id'] for s in control_stats if s['decision']=='failed_controls'}
    seen = defaultdict(set)
    control_ids = {s['review_id'] for s in controls}
    for row in first_rows:
        if pick(row,'ASSIGNMENT:status','assignment_status').upper() not in ('ACCEPTED','SUBMITTED'):continue
        rid=pick(row,'INPUT:review_id','review_id')
        if rid in control_ids:seen[rid].add(pick(row,'ASSIGNMENT:worker_id','worker_id'))
    possible = set(exam_passed)-set(excluded)-bad
    # Не повторяем знакомые эталоны. Минимум пять незнакомых на каждого кандидата.
    eligible = sorted(w for w in possible if sum(w not in seen[c] for c in control_ids)>=5)
    records=[]
    for q in queue:
        if q['issue']!='failed_controls':continue
        old=source[q['review_id']]
        previous=sorted({pick(r,'ASSIGNMENT:worker_id','worker_id') for r in q['raw_reviews']}-{''})
        records.append(dict(old,review_id=old['review_id']+'-recheck',original_review_id=old['review_id'],
                            previous_reviewers=previous))
    if len({r['review_id'] for r in records})!=len(records):raise ValueError('Повтор в партии')
    return dict(records=records,eligible=eligible,bad_workers=sorted(bad),
                seen_controls={k:sorted(v) for k,v in seen.items()},excluded=sorted(set(excluded)))

def rereview_payloads(plan, pool_id, controls):
    eligible=plan['eligible'];bad=set(plan['bad_workers'])|set(plan['excluded'])
    tasks=[]
    for r in plan['records']:
        forbidden=bad|{r['author_id']}|set(r['previous_reviewers'])
        allowed=sorted(set(eligible)-forbidden)
        if len(allowed)<2:raise ValueError('Для каждой работы нужны хотя бы два допустимых проверяющих')
        tasks.append(dict(pool_id=pool_id,input_values={k:r[k] for k in ('review_id','photo_id','image','boxes')},
            overlap=2,infinite_overlap=False,reserved_for=allowed,unavailable_for=sorted(forbidden)))
    for c in controls:
        forbidden=bad|set(plan['seen_controls'].get(c['review_id'],[]))
        allowed=sorted(set(eligible)-forbidden)
        if not allowed:continue
        tasks.append(dict(pool_id=pool_id,input_values={k:c[k] for k in ('review_id','photo_id','image','boxes')},
            known_solutions=[{'output_values':{'verdict':c['verdict']},'correctness_weight':1}],
            overlap=len(allowed),infinite_overlap=False,reserved_for=allowed,unavailable_for=sorted(forbidden)))
    return tasks

def create_rereview(plan, controls):
    state=read_rereview_state()
    if state.get('pool_id'):return state
    if state.get('creation_pending'):raise RuntimeError('Проверьте результат прошлого POST пула; автоматический повтор запрещён')
    if not plan['records']:return {}
    if len(plan['eligible'])<2:raise ValueError('Недостаточно новых квалифицированных ревьюеров с незнакомыми контролями')
    rereview_payloads(plan,'validation',controls)
    template=api_request('GET','/api/v1/pools/'+REVIEW_POOL_ID)
    payload={k:copy.deepcopy(template[k]) for k in ['project_id','will_expire','assignment_max_duration_seconds',
             'auto_accept_solutions','auto_accept_period_day','filter','quality_control'] if k in template}
    # Цена фиксирована по согласованному текущему тарифу, перекрытие реальных работ 2.
    payload.update(private_name='Повторный review — две независимые проверки',reward_per_assignment=0.2,
        defaults={'default_overlap_for_new_tasks':2,'default_overlap_for_new_task_suites':2},
        mixer_config={'real_tasks_count':1,'golden_tasks_count':0,'training_tasks_count':0,'min_real_tasks_count':1,
          'force_last_assignment':True,'shuffle_tasks_in_task_suite':True,'mix_tasks_in_creation_order':False,
          'golden_task_distribution_function':{'scope':'POOL','distribution':'UNIFORM','window_days':7,
              'intervals':[{'from':1,'to':5,'frequency':1},{'from':6,'to':1000000,'frequency':10}]}})
    for cfg in payload.get('quality_control',{}).get('configs',[]):cfg['collector_config'].pop('uuid',None)
    state={'creation_pending':True,'plan':plan,'created_at':datetime.now(timezone.utc).isoformat()}
    write_rereview_state(state)
    saved=api_request('POST','/api/v1/pools',payload)
    state.update(pool_id=str(saved['id']),creation_pending=False);write_rereview_state(state)
    if api_request('GET','/api/v1/pools/'+state['pool_id'])['status']!='CLOSED':raise RuntimeError('Новый пул не закрыт')
    return state

def aggregate_rereview(rows, plan, control_key, exam_passed, first_rows, confirmed):
    records={r['review_id']:r for r in plan['records']}
    active=[r for r in rows if pick(r,'ASSIGNMENT:status','assignment_status').upper() in ('SUBMITTED','ACCEPTED')]
    unknown={pick(r,'INPUT:review_id','review_id') for r in active}-set(records)-set(control_key)
    if unknown:raise ValueError('Неизвестные задания повторного review')
    first_controls=[r for r in first_rows if pick(r,'INPUT:review_id','review_id') in control_key]
    new_controls=[r for r in active if pick(r,'INPUT:review_id','review_id') in control_key]
    # Оцениваем и историю первого пула, и новые контроли. Дубли эталонов не засчитываем.
    _,_,all_stats=finalize_reviews(first_controls+new_controls,{},control_key,exam_passed,(),confirmed)
    _,_,new_stats=finalize_reviews(new_controls,{},control_key,exam_passed,(),confirmed)
    all_stats={r['worker_id']:r for r in all_stats};new_stats={r['worker_id']:r for r in new_stats}
    decisions=defaultdict(list)
    for row in active:
        rid=pick(row,'INPUT:review_id','review_id')
        if rid in records:decisions[rid].append(row)
    accepted,rejected,unresolved=[],[],[]
    for rid,source in records.items():
        answers=decisions[rid];workers=[pick(r,'ASSIGNMENT:worker_id','worker_id') for r in answers]
        issue=None
        if len(answers)!=2:issue='missing_rechecks' if len(answers)<2 else 'too_many_rechecks'
        elif len(set(workers))!=2:issue='same_reviewer_twice'
        elif any(w not in plan['eligible'] or w in plan['bad_workers'] or w in plan['excluded']
                 or w==source['author_id'] or w in source['previous_reviewers'] for w in workers):issue='reviewer_excluded'
        elif any(w not in exam_passed for w in workers):issue='exam_not_passed'
        elif any(all_stats.get(w,{}).get('decision')!='passed' or new_stats.get(w,{}).get('decision')!='passed' for w in workers):issue='reviewer_quality_not_confirmed'
        elif any(not pick(r,'ASSIGNMENT:assignment_id','assignment_id') or not pick(r,'ASSIGNMENT:submitted','submitted') for r in answers):issue='missing_provenance'
        else:
            votes=[pick(r,'OUTPUT:verdict','verdict') for r in answers]
            if votes==['accept','accept']:
                accepted.append(dict(source,verdict='accept',reviewer_ids=workers,
                    review_assignment_ids=[pick(r,'ASSIGNMENT:assignment_id','assignment_id') for r in answers]));continue
            if votes==['reject','reject']:
                rejected.append(dict(review_id=source['original_review_id'],photo_id=source['photo_id'],issue='reject',raw_reviews=answers,recheck_id=rid));continue
            issue='image_error' if 'image_error' in votes else 'disagreement_or_uncertain'
        unresolved.append(dict(review_id=rid,photo_id=source['photo_id'],issue=issue,raw_reviews=answers))
    return accepted,rejected,unresolved,list(new_stats.values())
```

## Независимый review переразметки — перекрытие 2
1. Ячейка выше скачивает новые рамки и сохраняет исходные назначения. Дожидаемся всех 95 корректных ответов.

4. После выполнения повторите выгрузку review и итоговую ячейку. Два надёжных accept принимают работу, два reject возвращают её на исправление. Расхождения accept/reject сохраняются в `corrections/review_overlap2/third_review_tasks.json`, с исключением первых двух ревьюеров. Это заготовка для отдельного третьего review, не автоматическая загрузка. Неизвестные/технические ответы и неподтверждённое качество не считаются голосами большинства.
Запас контролей конечен: при его исчерпании нужны новые проверенные эталоны. Не включайте старую загрузку v2 в основной пул с перекрытием 1.


```python
# Review новых рамок: две независимые проверки, без reserved_for.
DUAL_DIR = CORRECTIONS/'review_overlap2'; DUAL_DIR.mkdir(exist_ok=True)
DUAL_STATE = DUAL_DIR/'state.json'

def save_dual_state(state):
    temp=DUAL_STATE.with_suffix('.tmp')
    temp.write_text(json.dumps(state,ensure_ascii=False,indent=2));temp.replace(DUAL_STATE)

def dual_review_payloads(records, controls, seen, excluded, passed):
    blocked=set(excluded)
    known_workers=set().union(*seen.values()) if seen else set()
    # Первые пять контролей должны быть незнакомыми каждому участнику.
    blocked|={w for w in known_workers if sum(w not in seen.get(c['review_id'],set()) for c in controls)<5}
    eligible=set(passed)-blocked
    tasks=[]
    for r in records:
        unavailable=blocked|{r['author_id'],r['original_author_id']}|set(r['previous_reviewers'])
        if len(eligible-unavailable)<2:raise ValueError('Недостаточно независимых кандидатов с незнакомыми контролями')
        tasks.append(dict(input_values={k:r[k] for k in ('review_id','photo_id','image','boxes')},
                          overlap=2,infinite_overlap=False,unavailable_for=sorted(unavailable)))
    for c in controls:
        unavailable=blocked|set(seen.get(c['review_id'],set()))
        count=len(eligible-unavailable)
        if count:
            tasks.append(dict(input_values={k:c[k] for k in ('review_id','photo_id','image','boxes')},
                known_solutions=[{'output_values':{'verdict':c['verdict']},'correctness_weight':1}],
                overlap=count,infinite_overlap=False,unavailable_for=sorted(unavailable)))
    return tasks

def prepare_dual_review(records, controls):
    if len({r['review_id'] for r in records})!=len(records):raise ValueError('Повтор новой версии')
    if not controls_confirmed:raise ValueError('Нужны проверенные эталоны review')
    seen=defaultdict(set);history=[]
    prior=read_rereview_state()
    pool_ids=[REVIEW_POOL_ID]+([prior['pool_id']] if prior.get('pool_id') else [])+prior.get('continuation_pool_ids',[])
    for pid in dict.fromkeys(pool_ids):
        assignments=api_list('assignments',pool_id=pid)
        for a in assignments:
            for t in a['tasks']:
                rid=t['input_values']['review_id']
                if rid in control_key:seen[rid].add(a['user_id'])
        history.extend(r for r in fetch_review_rows(pid) if pick(r,'INPUT:review_id','review_id') in control_key)
    passed={s['user_id'] for s in api_list('user-skills',skill_id=REVIEW_SKILL_ID) if s['value']>=80}
    _,_,history_stats=finalize_reviews(history,{},control_key,passed,(),True)
    bad={s['worker_id'] for s in history_stats if s['decision'] not in ('passed','insufficient_controls')}
    excluded=set(authors)|{r['author_id'] for r in records}|bad
    excluded|={s['user_id'] for s in api_list('user-skills',skill_id='21809')}
    tasks=dual_review_payloads(records,controls,seen,excluded,passed)
    plan=dict(records=records,tasks=tasks,excluded=sorted(excluded),history_pool_ids=pool_ids,price=0.3)
    (DUAL_DIR/'plan.json').write_text(json.dumps(plan,ensure_ascii=False,indent=2))
    print('Фото:',len(records),'обычных проверок:',2*len(records),'цена: 0,30 ₽')
    print('Обычные проверки без комиссии:',round(2*len(records)*.3,2),'₽; контроли оплачиваются дополнительно.')
    return plan

def create_dual_review(plan):
    state=json.loads(DUAL_STATE.read_text()) if DUAL_STATE.exists() else {}
    if state.get('pool_id'):return state
    if state.get('creation_pending'):raise RuntimeError('Неясный результат POST пула — проверьте журнал')
    template=api_request('GET','/api/v1/pools/'+REVIEW_POOL_ID)
    payload={k:copy.deepcopy(template[k]) for k in ['project_id','will_expire','assignment_max_duration_seconds',
             'auto_accept_solutions','auto_accept_period_day','filter','quality_control'] if k in template}
    from datetime import timedelta
    payload.update(private_name='Review переразметки — две независимые проверки',reward_per_assignment=plan['price'],
        will_expire=(datetime.now(timezone.utc)+timedelta(days=14)).strftime('%Y-%m-%dT%H:%M:%S'),
        defaults={'default_overlap_for_new_tasks':2,'default_overlap_for_new_task_suites':2},
        mixer_config={'real_tasks_count':1,'golden_tasks_count':0,'training_tasks_count':0,'min_real_tasks_count':1,
         'force_last_assignment':True,'shuffle_tasks_in_task_suite':True,
         'golden_task_distribution_function':{'scope':'POOL','distribution':'UNIFORM','window_days':7,
          'intervals':[{'from':1,'to':5,'frequency':1},{'from':6,'to':1000000,'frequency':10}]}})
    for cfg in payload.get('quality_control',{}).get('configs',[]):cfg['collector_config'].pop('uuid',None)
    state={'creation_pending':True,'plan':plan};save_dual_state(state)
    try:
        saved=api_request('POST','/api/v1/pools',payload)
    except RuntimeError as error:
        # Явный отказ валидации не создаёт объект. Сетевой сбой оставляет pending.
        if getattr(error,'status_code',None)==400 and getattr(error,'validation_errors',None):
            state.update(creation_pending=False,last_validation_error=error.validation_errors)
            save_dual_state(state)
        raise
    state.update(pool_id=str(saved['id']),creation_pending=False);save_dual_state(state)
    return state

def aggregate_dual_review(rows, records, key, passed, excluded):
    source={r['review_id']:r for r in records}
    complete=[r for r in rows if pick(r,'ASSIGNMENT:status','assignment_status').upper() in ('ACCEPTED','SUBMITTED')]
    if {pick(r,'INPUT:review_id','review_id') for r in complete}-set(source)-set(key):raise ValueError('Неизвестные review_id')
    gold=[r for r in complete if pick(r,'INPUT:review_id','review_id') in key]
    _,_,stats=finalize_reviews(gold,{},key,passed,(),True)
    quality={s['worker_id']:s['decision'] for s in stats}
    groups=defaultdict(list)
    for r in complete:
        rid=pick(r,'INPUT:review_id','review_id')
        if rid in source:groups[rid].append(r)
    accepted=[];rejected=[];pending=[];third=[]
    for rid,record in source.items():
        votes=groups[rid];workers=[pick(r,'ASSIGNMENT:worker_id','worker_id') for r in votes]
        forbidden=set(excluded)|{record['author_id'],record['original_author_id']}|set(record['previous_reviewers'])
        issue=None
        if len(votes)!=2:issue='missing_reviews' if len(votes)<2 else 'too_many_reviews'
        elif len(set(workers))!=2:issue='same_reviewer_twice'
        elif any(w in forbidden or w not in passed for w in workers):issue='reviewer_excluded_or_not_qualified'
        elif any(quality.get(w)!='passed' for w in workers):issue='reviewer_quality_not_confirmed'
        elif any(not pick(r,'ASSIGNMENT:assignment_id','assignment_id') or not pick(r,'ASSIGNMENT:submitted','submitted') for r in votes):issue='missing_provenance'
        else:
            verdicts=[pick(r,'OUTPUT:verdict','verdict') for r in votes]
            if verdicts==['accept','accept']:
                accepted.append(dict(record,verdict='accept',reviewer_ids=workers,review_assignment_ids=[pick(r,'ASSIGNMENT:assignment_id','assignment_id') for r in votes]));continue
            if verdicts==['reject','reject']:
                rejected.append(dict(review_id=rid,photo_id=record['photo_id'],issue='reject',raw_reviews=votes));continue
            issue='disagreement' if set(verdicts)=={'accept','reject'} else 'image_error_or_uncertain'
            if issue=='disagreement':
                third.append(dict(input_values={k:record[k] for k in ('review_id','photo_id','image','boxes')},
                    overlap=1,infinite_overlap=False,unavailable_for=sorted(forbidden|set(workers))))
        pending.append(dict(review_id=rid,photo_id=record['photo_id'],issue=issue,raw_reviews=votes))
    return accepted,rejected,pending,third,stats
```

## Финальный этап: 22 фото × 2 разметчика → сравнение → review вариантов
Цель: получить две независимые новые разметки каждой незакрытой фотографии. Обе исходные версии сохраняются. «Одинаково» означает одинаковое число объектов и взаимно-однозначное сопоставление рамок с IoU ≥0,85; два корректных пустых ответа также совпадают. При совпадении на review отправляется первый вариант по ID назначения (не усреднённые рамки). При расхождении проверяются оба варианта отдельно. Ошибка загрузки никогда не считается совпадением.

Каждый вариант проверяют два квалифицированных ревьюера с контролями. Два accept подтверждают вариант, два reject отклоняют; разногласия и неподтверждённое качество остаются нерешёнными. Если оба разных варианта подтверждены — сохраняем оба, в итоговый набор выбираем первый по ID; это не утверждение, что второй неправильный. Для принятия фото достаточно одного варианта с двумя квалифицированными accept. Разногласие по альтернативному варианту не отменяет это подтверждение. Если подтверждены оба варианта, выбор представителя детерминированный, а не оценка превосходства одного над другим.

Пулы без reserved_for. Из переразметки исключены прежние авторы и проверяющие; из review — авторы и предыдущие проверяющие. Контроли не повторяются для видевших их исполнителей. Нужны минимум пять свежих контролей и качество ≥80%.

Бюджет: 44 разметки ×0,20 =8,80 ₽; 44–88 review ×0,30 =13,20–26,40 ₽. При комиссии 40%: 30,80–49,28 ₽ без контролей. Контроли — максимум 14×5=70 ответов, ещё до29,40 ₽ с комиссией. Плановый максимум одной полной выдачи: 78,68 ₽; повторные выдачи и дальнейший разбор в него не входят. Проверенный остаток перед этапом: 250,66 ₽.

Все действия воспроизводимы следующими ячейками. Создание и открытие разделены. При неполной/некорректной разметке создание review блокируется. Исходные назначения, сравнения, решения и итоговые версии сохраняются в corrections/final_stage/. Экспорт ready/final_accepted.json содержит только подтверждённые версии; final_remaining.json — незакрытые фото.


```python
# Финальный этап: две разметки каждого фото, затем review одного или обоих вариантов.
FINAL_DIR=CORRECTIONS/'final_stage';FINAL_DIR.mkdir(exist_ok=True)
FINAL_STATE=FINAL_DIR/'state.json'
FINAL_EQUIVALENCE_IOU=0.85
FINAL_CONTROL_OVERLAP=5

def final_write(path,value):
    temp=path.with_suffix('.tmp');temp.write_text(json.dumps(value,ensure_ascii=False,indent=2));temp.replace(path)

def final_state():
    return json.loads(FINAL_STATE.read_text()) if FINAL_STATE.exists() else {}

def final_create_pool(key,payload):
    st=final_state()
    if st.get(key):return st[key]
    if st.get('pending'):raise RuntimeError('Неясный результат создания: проверьте final_stage/state.json')
    st['pending']=key;final_write(FINAL_STATE,st)
    try:saved=api_request('POST','/api/v1/pools',payload)
    except RuntimeError as error:
        if getattr(error,'status_code',None)==400 and getattr(error,'validation_errors',None):
            st.pop('pending');final_write(FINAL_STATE,st)
        raise
    st[key]=str(saved['id']);st.pop('pending');final_write(FINAL_STATE,st);return st[key]

def final_bbox_create():
    st=final_state()
    if not st.get('batch'):
        batch=json.loads((DUAL_DIR/'next_relabel_batch.json').read_text())
        if not batch:raise ValueError('Нет незакрытых работ')
        if len({r['photo_id'] for r in batch})!=len(batch):raise ValueError('Дубли фото')
        st['batch']=batch;st['equivalence_iou']=FINAL_EQUIVALENCE_IOU;final_write(FINAL_STATE,st)
    template=api_request('GET','/api/v1/pools/'+load_correction_state()['pool_id'])
    payload={k:copy.deepcopy(template[k]) for k in ['project_id','filter','quality_control','assignment_max_duration_seconds','auto_accept_solutions','auto_accept_period_day'] if k in template}
    from datetime import timedelta
    payload.update(private_name='Финальные исправления — две независимые разметки',reward_per_assignment=.2,
        will_expire=(datetime.now(timezone.utc)+timedelta(days=14)).strftime('%Y-%m-%dT%H:%M:%S'),
        defaults={'default_overlap_for_new_tasks':2,'default_overlap_for_new_task_suites':2},
        mixer_config={'real_tasks_count':1,'golden_tasks_count':0,'training_tasks_count':0,'min_real_tasks_count':1,'force_last_assignment':True})
    for cfg in payload.get('quality_control',{}).get('configs',[]):cfg['collector_config'].pop('uuid',None)
    pid=final_create_pool('bbox_pool_id',payload)
    tasks=[dict(pool_id=pid,input_values={'photo_id':r['photo_id'],'image':r['image']},overlap=2,infinite_overlap=False,
        unavailable_for=sorted({r['original_author_id'],r['previous_author_id']}|set(r['previous_reviewers']))) for r in st['batch']]
    upload_correction_tasks(pid,tasks,'final_bbox_upload.json')
    return pid

def matching_boxes(left,right,threshold=.85):
    if len(left)!=len(right):return False
    def iou(a,b):
        x=max(0,min(a['left']+a['width'],b['left']+b['width'])-max(a['left'],b['left']))
        y=max(0,min(a['top']+a['height'],b['top']+b['height'])-max(a['top'],b['top']))
        intersection=x*y;union=a['width']*a['height']+b['width']*b['height']-intersection
        return intersection/union if union>0 else 0
    edges=[[j for j,b in enumerate(right) if iou(a,b)+1e-9>=threshold] for a in left]
    owners={}
    def match(i,seen):
        for j in edges[i]:
            if j in seen:continue
            seen.add(j)
            if j not in owners or match(owners[j],seen):owners[j]=i;return True
        return False
    return all(match(i,set()) for i in range(len(left)))

def final_variants(assignments,state):
    batch={r['photo_id']:r for r in state['batch']};by_photo=defaultdict(list);issues=[]
    for a in assignments:
        if str(a['pool_id'])!=state['bbox_pool_id']:raise ValueError('Чужой bbox-пул')
        if a['status'] not in ('ACCEPTED','SUBMITTED'):continue
        if len(a['tasks'])!=len(a['solutions']):raise ValueError('Не совпадает количество задач и ответов')
        solutions=a['solutions']
        if any(s.get('task_id') for s in solutions):
            mapping={s['task_id']:s for s in solutions};solutions=[mapping[t['id']] for t in a['tasks']]
        for t,sol in zip(a['tasks'],solutions):
            photo=t['input_values']['photo_id']
            if photo not in batch:raise ValueError('Неизвестное фото')
            by_photo[photo].append((a,t,sol))
    variants=[];comparisons=[]
    for photo,old in batch.items():
        answers=sorted(by_photo[photo],key=lambda v:v[0]['id'])
        if len(answers)!=2:
            issues.append(dict(photo_id=photo,issue='missing_annotations' if len(answers)<2 else 'too_many_annotations'));continue
        if len({a[0]['user_id'] for a in answers})!=2:
            issues.append(dict(photo_id=photo,issue='same_author_twice'));continue
        records=[]
        for index,(a,t,sol) in enumerate(answers):
            forbidden={old['original_author_id'],old['previous_author_id']}|set(old['previous_reviewers'])
            try:
                if a['user_id'] in forbidden:raise ValueError('excluded_author')
                if t['input_values']['image']!=old['image']:raise ValueError('changed_image')
                out=sol['output_values'];boxes=boxes_ltwh(out.get('result'))
                if out.get('status') not in ('boxes','none') or (out['status']=='none')!=(len(boxes)==0):raise ValueError('image_error_or_invalid_answer')
                records.append(dict(review_id='review-'+str(photo)+'-v3-'+str(index+1),photo_id=photo,image=old['image'],boxes=boxes,
                    author_id=a['user_id'],assignment_id=a['id'],attempt=3,original_author_id=old['original_author_id'],
                    previous_review_id=old['previous_review_id'],previous_reviewers=sorted(set(old['previous_reviewers'])|{old['previous_author_id']})))
            except (KeyError,ValueError,TypeError) as error:issues.append(dict(photo_id=photo,issue=str(error),assignment_id=a['id']))
        if len(records)!=2:continue
        equivalent=matching_boxes(records[0]['boxes'],records[1]['boxes'],state['equivalence_iou'])
        # Representative is deterministic by assignment ID, never an average or model-generated box.
        selected=records[:1] if equivalent else records
        comparisons.append(dict(photo_id=photo,equivalent=equivalent,iou_threshold=state['equivalence_iou'],
            original_variants=records,review_ids=[r['review_id'] for r in selected]))
        variants.extend(selected)
    return variants,comparisons,issues

def final_review_plan(records):
    if not controls_confirmed:raise ValueError('Нужны проверенные эталоны review')
    if not records:raise ValueError('Нет вариантов для проверки')
    previous=read_rereview_state();pool_ids=[REVIEW_POOL_ID,previous.get('pool_id'),*previous.get('continuation_pool_ids',[])]
    if DUAL_STATE.exists():pool_ids.append(json.loads(DUAL_STATE.read_text()).get('pool_id'))
    seen=defaultdict(set);history=[]
    for pid in dict.fromkeys(p for p in pool_ids if p):
        for a in api_list('assignments',pool_id=pid):
            for t in a['tasks']:
                rid=t['input_values']['review_id']
                if rid in control_key:seen[rid].add(a['user_id'])
        history.extend(r for r in fetch_review_rows(pid) if pick(r,'INPUT:review_id','review_id') in control_key)
    passed={s['user_id'] for s in api_list('user-skills',skill_id=REVIEW_SKILL_ID) if s['value']>=80}
    _,_,stats=finalize_reviews(history,{},control_key,passed,(),True)
    excluded=set(authors)|{r['author_id'] for r in records}|{s['worker_id'] for s in stats if s['decision'] not in ('passed','insufficient_controls')}
    excluded|={s['user_id'] for s in api_list('user-skills',skill_id='21809')}
    tasks=dual_review_payloads(records,controls,seen,excluded,passed)
    for task in tasks:
        if task.get('known_solutions'):task['overlap']=min(task['overlap'],FINAL_CONTROL_OVERLAP)
    return dict(records=records,tasks=tasks,excluded=sorted(excluded),price=.3)

def final_review_create(plan):
    st=final_state()
    if not st.get('review_pool_id'):
        latest_variants=json.loads((FINAL_DIR/'variants.json').read_text())
        expected={r['review_id']:r for r in latest_variants}
        supplied={r['review_id']:r for r in plan['records']}
        if expected!=supplied:
            raise ValueError('План review устарел относительно variants.json. Повторите Финал 2 и Финал 3; не загружайте старую переменную final_plan.')
    if st.get('review_plan') and st['review_plan']!=plan:raise ValueError('План уже зафиксирован, нельзя заменить варианты')
    st['review_plan']=plan;final_write(FINAL_STATE,st)
    template=api_request('GET','/api/v1/pools/'+json.loads(DUAL_STATE.read_text())['pool_id'])
    payload={k:copy.deepcopy(template[k]) for k in ['project_id','filter','quality_control','assignment_max_duration_seconds','auto_accept_solutions','auto_accept_period_day','mixer_config','defaults'] if k in template}
    from datetime import timedelta
    payload.update(private_name='Финальный review — проверка вариантов рамок',reward_per_assignment=.3,
        will_expire=(datetime.now(timezone.utc)+timedelta(days=14)).strftime('%Y-%m-%dT%H:%M:%S'))
    for cfg in payload.get('quality_control',{}).get('configs',[]):cfg['collector_config'].pop('uuid',None)
    pid=final_create_pool('review_pool_id',payload)
    upload_correction_tasks(pid,[dict(t,pool_id=pid) for t in plan['tasks']],'final_review_upload.json')
    return pid

def select_final_versions(comparisons,accepted,rejected,unresolved):
    good={r['review_id']:r for r in accepted};bad={r['review_id'] for r in rejected};pending={r['review_id'] for r in unresolved}
    selected=[];queue=[]
    for photo in comparisons:
        ids=photo['review_ids'];ready=[good[rid] for rid in ids if rid in good]
        if ready:
            selected.append(dict(sorted(ready,key=lambda r:r['review_id'])[0],
                accepted_variant_ids=[r['review_id'] for r in ready],comparison_equivalent=photo['equivalent']))
        elif any(rid in pending or rid not in good and rid not in bad for rid in ids):
            queue.append(dict(photo_id=photo['photo_id'],issue='variant_review_unresolved',review_ids=ids))
        else:queue.append(dict(photo_id=photo['photo_id'],issue='all_variants_rejected',review_ids=ids))
    return selected,queue

def final_open_pool(pool_id):
    pool=api_request('GET','/api/v1/pools/'+pool_id)
    if pool['status']=='OPEN':return
    warnings=api_request('GET','/api/new/requester/pools/'+pool_id+'/validate')
    blockers=[r['type'] for r in warnings if r['active'] and r['blocker']]
    if blockers:raise ValueError('Блокеры запуска: '+str(blockers))
    operation=api_request('POST','/api/v1/pools/'+pool_id+'/open')
    import time
    for _ in range(30):
        result=api_request('GET','/api/v1/operations/'+operation['id'])
        if result['status']=='SUCCESS':
            if api_request('GET','/api/v1/pools/'+pool_id)['status']!='OPEN':raise RuntimeError('Пул не открыт')
            return
        if result['status']=='FAIL':raise RuntimeError('Ошибка запуска пула')
        time.sleep(1)
    raise RuntimeError('Операция ещё выполняется; проверьте статус перед повтором')
```

```python
# Диагностика различий: не меняет рамки, порог или план review.
comparison_diagnostic=[]
comparisons_for_diagnostic=json.loads((FINAL_DIR/'comparisons.json').read_text())
def diagnostic_iou(a,b):
    w=max(0,min(a['left']+a['width'],b['left']+b['width'])-max(a['left'],b['left']))
    h=max(0,min(a['top']+a['height'],b['top']+b['height'])-max(a['top'],b['top']))
    intersection=w*h
    return intersection/(a['width']*a['height']+b['width']*b['height']-intersection)
for item in comparisons_for_diagnostic:
    a,b=[r['boxes'] for r in item['original_variants']]
    if not a and not b:
        comparison_diagnostic.append(dict(photo_id=item['photo_id'],empty=True,iou=1,shifted_iou=1,shift_2pct_iou=1,shift_5pct_iou=1,center_dx=0,center_dy=0,max_edge_diff=0));continue
    if len(a)!=1 or len(b)!=1:
        comparison_diagnostic.append(dict(photo_id=item['photo_id'],empty=False,issue='requires_multiobject_matching'));continue
    a,b=a[0],b[0]
    dx=(a['left']+a['width']/2)-(b['left']+b['width']/2)
    dy=(a['top']+a['height']/2)-(b['top']+b['height']/2)
    def shifted(limit):
        sx=max(-limit,min(limit,dx));sy=max(-limit,min(limit,dy))
        # Translation only, keeping width/height and the whole rectangle inside the image.
        x=max(0,min(1-b['width'],b['left']+sx));y=max(0,min(1-b['height'],b['top']+sy))
        return diagnostic_iou(a,dict(b,left=x,top=y))
    comparison_diagnostic.append(dict(photo_id=item['photo_id'],empty=False,iou=diagnostic_iou(a,b),
        shifted_iou=shifted(1),shift_2pct_iou=shifted(.02),shift_5pct_iou=shifted(.05),
        center_dx=dx,center_dy=dy,width_ratio=b['width']/a['width'],height_ratio=b['height']/a['height'],
        area_ratio=max(a['width']*a['height'],b['width']*b['height'])/min(a['width']*a['height'],b['width']*b['height']),
        max_edge_diff=max(abs(a['left']-b['left']),abs(a['top']-b['top']),
            abs(a['left']+a['width']-b['left']-b['width']),abs(a['top']+a['height']-b['top']-b['height']))))
final_write(FINAL_DIR/'similarity_diagnostic.json',comparison_diagnostic)
print('Порог IoU | Исходные совпадения | После сдвига до 2% по каждой оси | до 5% | Совмещение центров')
for threshold in [.95,.9,.85,.8,.75,.5]:
    counts=[sum(r.get(k,-1)+1e-9>=threshold for r in comparison_diagnostic) for k in ['iou','shift_2pct_iou','shift_5pct_iou','shifted_iou']]
    print(threshold,*counts)
print('Числа включают два пустых совпадения; сдвиги — только диагностические, исходные ответы не менялись.')
print('Фото с наименьшим IoU:')
for r in sorted(comparison_diagnostic,key=lambda r:r.get('iou',-1))[:10]:
    print(r['photo_id'],'IoU',round(r.get('iou',0),3),'после совмещения центров',round(r.get('shifted_iou',0),3),
          'отношение площадей',round(r.get('area_ratio',1),2),'максимальная разница границы',round(100*r.get('max_edge_diff',0),1),'% изображения')
```

```python
# Финал 5. Скачать review, выбрать подтверждённую версию и собрать финальный набор.
st=final_state();final_selected=[];final_pending=[]
if st.get('review_pool_id'):
    rows=fetch_review_rows(st['review_pool_id']);save_review_export(FINAL_DIR/'review_results.tsv',rows)
    passed={s['user_id'] for s in api_list('user-skills',skill_id=REVIEW_SKILL_ID) if s['value']>=80}
    yes,no,pending,third,stats_final=aggregate_dual_review(rows,st['review_plan']['records'],control_key,passed,st['review_plan']['excluded'])
    # Выбор ведётся по реально загруженным вариантам, даже если локальный порог позже менялся.
    actual_groups=defaultdict(list)
    for record in st['review_plan']['records']:actual_groups[record['photo_id']].append(record['review_id'])
    local_comparisons={r['photo_id']:r for r in final_comparisons}
    selection_comparisons=[dict(photo_id=photo,review_ids=sorted(ids),
        equivalent=local_comparisons.get(photo,{}).get('equivalent',len(ids)==1)) for photo,ids in actual_groups.items()]
    final_selected,final_pending=select_final_versions(selection_comparisons,yes,no,pending)
    for name,data in [('accepted_variants',yes),('rejected_variants',no),('unresolved_variants',pending),('control_scores',stats_final),('selected',final_selected)]:
        final_write(FINAL_DIR/(name+'.json'),data)
    print('Подтверждённых вариантов:',len(yes),'отклонённых:',len(no),'ожидание/разбор:',len(pending))
    print('Выбрана версия для фото:',len(final_selected))
if st.get('batch'):
    batch_ids={r['photo_id'] for r in st['batch']}
    existing=json.loads((OUT/'accepted_latest.json').read_text())
    final_dataset=[r for r in existing if r['photo_id'] not in batch_ids]+final_selected
    selected_ids={r['photo_id'] for r in final_selected};pending_map={r['photo_id']:r for r in final_pending}
    unresolved_final=[pending_map.get(r['photo_id'],dict(photo_id=r['photo_id'],issue='final_stage_pending')) for r in st['batch'] if r['photo_id'] not in selected_ids]
    if len({r['photo_id'] for r in final_dataset})!=len(final_dataset):raise ValueError('Дубли фото в финальном наборе')
    if {r['photo_id'] for r in final_dataset}&{r['photo_id'] for r in unresolved_final}:raise ValueError('Фото одновременно принято и не закрыто')
    if len(final_dataset)+len(unresolved_final)!=len(original_sources):raise ValueError('Потеря фото при сборке')
    final_write(OUT/'final_accepted.json',final_dataset);final_write(OUT/'final_remaining.json',unresolved_final)
    print('Итого принято:',len(final_dataset),'осталось:',len(unresolved_final),'из',len(original_sources))
```

Цена 0,30 ₽ за страницу, перекрытие 1. Первые пять выдач — незнакомые контроли, далее контроль раз в 10; в локальной оценке нужны минимум 5 контролей и точность не ниже 80%. Контроли оплачиваются дополнительно.
Принят любой подтверждённый вариант фото. Если все спорные варианты отклонены, фото остаётся в очереди; повторная переразметка автоматически не запускается.


```python
# Финал 7. Подготовка третьего независимого голоса по оставшимся вариантам.
st=final_state()
if st.get('review_pool_id') and (FINAL_DIR/'unresolved_variants.json').exists():
    remaining=json.loads((OUT/'final_remaining.json').read_text());remaining_ids={r['photo_id'] for r in remaining}
    source={r['review_id']:r for r in st['review_plan']['records']}
    unresolved=json.loads((FINAL_DIR/'unresolved_variants.json').read_text())
    adjudication_tasks=[]
    for item in unresolved:
        if item['photo_id'] not in remaining_ids or item['issue']!='disagreement':continue
        record=source[item['review_id']]
        old_reviewers={pick(r,'ASSIGNMENT:worker_id','worker_id') for r in item['raw_reviews']}-{''}
        forbidden=set(st['review_plan']['excluded'])|{record['author_id'],record['original_author_id']}|set(record['previous_reviewers'])|old_reviewers
        adjudication_tasks.append(dict(input_values={k:record[k] for k in ('review_id','photo_id','image','boxes')},
            overlap=1,infinite_overlap=False,unavailable_for=sorted(forbidden)))
    final_write(FINAL_DIR/'adjudication_tasks.json',adjudication_tasks)
    print('Осталось фото:',len(remaining_ids),'спорных вариантов для третьего голоса:',len(adjudication_tasks))
    print('Новый пул не создан. Перед загрузкой нужны незнакомые контроли и проверка допуска.')
```

```python
# Третий независимый review спорных вариантов. Все записи API выключены по умолчанию.
ADJ_DIR=FINAL_DIR/'adjudication';ADJ_DIR.mkdir(exist_ok=True)
ADJ_STATE=ADJ_DIR/'state.json'

def adjudication_plan():
    if ADJ_STATE.exists():return json.loads(ADJ_STATE.read_text())['plan']
    if not controls_confirmed:raise ValueError('Нужны проверенные контроли')
    st=final_state();remaining=json.loads((OUT/'final_remaining.json').read_text())
    wanted={r['photo_id'] for r in remaining}
    passed={s['user_id'] for s in api_list('user-skills',skill_id=REVIEW_SKILL_ID) if s['value']>=80}
    previous_rows=fetch_review_rows(st['review_pool_id'])
    _,_,pending,_,old_stats=aggregate_dual_review(previous_rows,st['review_plan']['records'],control_key,passed,st['review_plan']['excluded'])
    source={r['review_id']:r for r in st['review_plan']['records']}
    disputes=[r for r in pending if r['photo_id'] in wanted and r['issue']=='disagreement']
    if {r['photo_id'] for r in disputes}!=wanted:raise ValueError('Не все оставшиеся фото имеют спорные варианты')
    seen=defaultdict(set);history=[];prior=read_rereview_state()
    pools=list(dict.fromkeys([REVIEW_POOL_ID,prior['pool_id'],*prior.get('continuation_pool_ids',[]),json.loads(DUAL_STATE.read_text())['pool_id'],st['review_pool_id']]))
    for pid in pools:
        for a in api_list('assignments',pool_id=pid):
            for t in a['tasks']:
                rid=t['input_values']['review_id']
                if rid in control_key:seen[rid].add(a['user_id'])
        history.extend(r for r in fetch_review_rows(pid) if r['INPUT:review_id'] in control_key)
    _,_,history_stats=finalize_reviews(history,{},control_key,passed,(),True)
    blocked=set(st['review_plan']['excluded'])|{s['worker_id'] for s in history_stats if s['decision'] not in ('passed','insufficient_controls')}
    blocked|={s['user_id'] for s in api_list('user-skills',skill_id='21809')}
    blocked|={w for w in set().union(*seen.values()) if sum(w not in seen.get(c['review_id'],set()) for c in controls)<5}
    eligible=passed-blocked;previous_by_photo=defaultdict(set)
    for row in previous_rows:
        if row['INPUT:review_id'] in source:previous_by_photo[source[row['INPUT:review_id']]['photo_id']].add(row['ASSIGNMENT:worker_id'])
    records=[];tasks=[];candidate_counts=[]
    for item in disputes:
        rec=copy.deepcopy(source[item['review_id']]);rec['previous_reviewers']=sorted(set(rec['previous_reviewers'])|previous_by_photo[rec['photo_id']])
        forbidden=blocked|set(rec['previous_reviewers'])|{rec['author_id'],rec['original_author_id']}
        count=len(eligible-forbidden)
        if count<1:raise ValueError('Нет независимых кандидатов с 5 незнакомыми контролями для '+rec['review_id'])
        candidate_counts.append(count);records.append(rec)
        tasks.append(dict(input_values={k:rec[k] for k in ('review_id','photo_id','image','boxes')},overlap=1,infinite_overlap=False,unavailable_for=sorted(forbidden)))
    for c in controls:
        forbidden=blocked|seen.get(c['review_id'],set());count=min(5,len(eligible-forbidden))
        if count:tasks.append(dict(input_values={k:c[k] for k in ('review_id','photo_id','image','boxes')},known_solutions=[dict(output_values={'verdict':c['verdict']},correctness_weight=1)],overlap=count,infinite_overlap=False,unavailable_for=sorted(forbidden)))
    plan=dict(records=records,tasks=tasks,excluded=sorted(blocked),old_disputes=disputes,old_control_scores=old_stats,
        baseline_accepted=json.loads((OUT/'final_accepted.json').read_text()),baseline_remaining=remaining,price=.3,history_pool_ids=pools,
        min_candidates=min(candidate_counts),control_key=control_key)
    final_write(ADJ_DIR/'plan.json',plan)
    print('Фото:',len(wanted),'вариантов:',len(records),'минимум кандидатов до остальных фильтров:',plan['min_candidates'])
    print('Оценка всех выдач с контролями и комиссией 40%:',round(sum(t['overlap'] for t in tasks)*.3*1.4,2),'₽')
    return plan

def create_adjudication(plan):
    state=json.loads(ADJ_STATE.read_text()) if ADJ_STATE.exists() else {}
    if state and state['plan']!=plan:raise ValueError('План уже зафиксирован')
    if state.get('creation_pending'):raise RuntimeError('Неясный результат создания — проверьте журнал, не повторяйте POST')
    if not state.get('pool_id'):
        template=api_request('GET','/api/v1/pools/'+final_state()['review_pool_id'])
        payload={k:copy.deepcopy(template[k]) for k in ('project_id','filter','quality_control','assignment_max_duration_seconds','auto_accept_solutions','auto_accept_period_day','mixer_config') if k in template}
        from datetime import timedelta
        payload.update(private_name='Третий review — 9 спорных фото, большинство 2 из 3',reward_per_assignment=plan['price'],will_expire=(datetime.now(timezone.utc)+timedelta(days=14)).strftime('%Y-%m-%dT%H:%M:%S'),defaults={'default_overlap_for_new_tasks':1,'default_overlap_for_new_task_suites':1})
        for cfg in payload.get('quality_control',{}).get('configs',[]):cfg['collector_config'].pop('uuid',None)
        state=dict(plan=plan,creation_pending=True);final_write(ADJ_STATE,state)
        try:saved=api_request('POST','/api/v1/pools',payload)
        except RuntimeError as error:
            if getattr(error,'status_code',None)==400:
                state['creation_pending']=False;final_write(ADJ_STATE,state)
            raise
        state.update(pool_id=str(saved['id']),creation_pending=False);final_write(ADJ_STATE,state)
    pid=state['pool_id'];pool=api_request('GET','/api/v1/pools/'+pid)
    if pool['status']!='OPEN':upload_correction_tasks(pid,[dict(t,pool_id=pid) for t in plan['tasks']],'adjudication_upload.json')
    return pid

def aggregate_adjudication(rows,plan,passed):
    key=plan['control_key'];source={r['review_id']:r for r in plan['records']}
    complete=[r for r in rows if pick(r,'ASSIGNMENT:status','assignment_status') in ('ACCEPTED','SUBMITTED')]
    if {r['INPUT:review_id'] for r in complete}-set(source)-set(key):raise ValueError('Неизвестные задания')
    _,_,stats=finalize_reviews([r for r in complete if r['INPUT:review_id'] in key],{},key,passed,(),True)
    quality={r['worker_id']:r['decision'] for r in stats};old_quality={r['worker_id']:r['decision'] for r in plan['old_control_scores']}
    old={r['review_id']:r['raw_reviews'] for r in plan['old_disputes']};yes=[];no=[];pending=[]
    for rid,rec in source.items():
        votes=[r for r in complete if r['INPUT:review_id']==rid];prior=old[rid]
        old_workers={r['ASSIGNMENT:worker_id'] for r in prior}
        if len(prior)!=2 or len(old_workers)!=2 or {r['OUTPUT:verdict'] for r in prior}!={'accept','reject'} or any(old_quality.get(w)!='passed' for w in old_workers):raise ValueError('Неверная история двух голосов')
        issue=None
        if len(votes)!=1:issue='missing_third_review' if not votes else 'too_many_reviews'
        else:
            v=votes[0];worker=v['ASSIGNMENT:worker_id'];forbidden=set(plan['excluded'])|set(rec['previous_reviewers'])|old_workers|{rec['author_id'],rec['original_author_id']}
            if worker in forbidden or worker not in passed:issue='reviewer_excluded_or_not_qualified'
            elif quality.get(worker)!='passed':issue='reviewer_quality_not_confirmed'
            elif not v.get('ASSIGNMENT:assignment_id') or not v.get('ASSIGNMENT:submitted'):issue='missing_provenance'
            elif v['OUTPUT:verdict'] not in ('accept','reject'):issue='image_error_or_uncertain'
            else:
                all_votes=prior+votes
                if sum(r['OUTPUT:verdict']=='accept' for r in all_votes)>=2:
                    yes.append(dict(rec,verdict='accept',decision_rule='majority_2_of_3',reviewer_ids=[r['ASSIGNMENT:worker_id'] for r in all_votes],review_assignment_ids=[r['ASSIGNMENT:assignment_id'] for r in all_votes],raw_reviews=all_votes))
                else:no.append(dict(review_id=rid,photo_id=rec['photo_id'],issue='reject',raw_reviews=all_votes))
                continue
        pending.append(dict(review_id=rid,photo_id=rec['photo_id'],issue=issue,raw_reviews=prior+votes))
    return yes,no,pending,stats

def download_adjudication():
    if not ADJ_STATE.exists():print('Пул третьего review ещё не создан');return
    state=json.loads(ADJ_STATE.read_text());plan=state['plan'];rows=fetch_review_rows(state['pool_id'])
    save_review_export(ADJ_DIR/'results.tsv',rows)
    passed={s['user_id'] for s in api_list('user-skills',skill_id=REVIEW_SKILL_ID) if s['value']>=80}
    yes,no,pending,stats=aggregate_adjudication(rows,plan,passed)
    groups=defaultdict(list)
    for r in plan['records']:groups[r['photo_id']].append(r['review_id'])
    selected,remaining=select_final_versions([dict(photo_id=p,review_ids=rs,equivalent=False) for p,rs in groups.items()],yes,no,pending)
    dataset=plan['baseline_accepted']+selected
    if len({r['photo_id'] for r in dataset})!=len(dataset) or len(dataset)+len(remaining)!=len(plan['baseline_accepted'])+len(plan['baseline_remaining']):raise ValueError('Потеря или дубли фото')
    for name,value in [('accepted_variants',yes),('rejected_variants',no),('unresolved_variants',pending),('control_scores',stats),('selected',selected)]:final_write(ADJ_DIR/(name+'.json'),value)
    final_write(OUT/'final_accepted.json',dataset);final_write(OUT/'final_remaining.json',remaining)
    print('Принято:',len(dataset),'осталось:',len(remaining),'из',len(dataset)+len(remaining))
    print('Новых принятых фото:',len(selected),'контроль ревьюеров:',dict(Counter(s['decision'] for s in stats)))
    return dataset,remaining
```
